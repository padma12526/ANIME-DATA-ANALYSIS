```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
df=pd.read_csv("anime.csv")
```

```python
# FINDING EPISODES ADDING EXTRA COLUM
```

```python
df.loc[1]["Title"]
```

```python
def epi(txt):
    a=txt.find("(")
    b=txt.find(")")
    a=a+1
    l=txt[a:b]
    return l
```

```python
df["Episodes"]=df["Title"].apply(epi)
```

```python
df['Episodes']= df['Episodes'].str.replace(" eps","")
df.head(2) # removing the eps with blank
```

```python
def e(txt):
    a=txt.find(")")
    a=a+1
    l=txt[a:a+19]
    return l
```

```python
df["Total Time"]=df["Title"].apply(e)
```

```python
# TOTAL TIME NEW COLOUM
```

```python
def mon(x):
    a=x.split("-")
    start=pd.to_datetime(a[0])
    end=pd.to_datetime(a[1])
    result=(end.year-start.year)*12+end.month-start.month+1
    return result
mon("Apr 2009 - Jul 2010")
```

```python
df["Months"]=df["Total Time"].apply(mon)
```
